# Practical 4 - Understanding Data with Graphs

Aim:

To understand and analyse data using statistical techniques and graphical representations with Python libraries.

Objectives:

To explore the Teaching Ratings dataset.
To identify duplicate observations.
To compare mean age and standard deviation before and after removing duplicate observations.
To calculate average teaching evaluation scores by course division.
To visualize the relationship between age and teaching evaluation.
To analyse data based on gender and tenure using scatter plots.
Importing the required Python libraries. Pandas is used for data handling, NumPy for numerical operations, Matplotlib and Seaborn for creating graphs and visualizations.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)
url = "https://vincentarelbundock.github.io/Rdatasets/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)
print(df.shape)
df.head()
```

```python
dup= df[df.duplicated(subset='prof',keep=False)]
dup=dup.sort_values('prof')
dup
```

```python
print("No. of unique instructors:",df['prof'].nunique())
```

```python
mean_age=df['age'].mean()
print("average age:",mean_age)
```

```python
std_dev=df['age'].std()
print("standard deviation:",std_dev)
```

```python
unique_instructors=df.drop_duplicates(subset='prof')
print(unique_instructors.shape)
```

```python
mean_age_unique=unique_instructors['age'].mean()
print("average age:",mean_age_unique)
```

```python
stdage_unique=unique_instructors['age'].std()
print("standard deviation:",stdage_unique)
```

```python
division_eva= df.groupby('division')['eval'].mean()
print(division_eva)
```

```python
sns.barplot(data=df,x='division',y='eval')
plt.title('Average Evaluation Score by Division')
plt.xlabel('course division')
plt.ylabel('Evaluation Score')
plt.show()
```

yes, instructors teaching lower division courses receive a bit higher average teaching evaluations

```python
sns.scatterplot(data=df,x='age',y='eval')
plt.title('Average Evaluation Score by Age')
plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.show()
```

this indicates that there is no strong relation between instructors' age and teaching eval. score. the eval. scores are across all age groups, this suggests yhat age is not a major factor determining teaching evaluations

```python
sns.scatterplot(data=df,x='age',y='eval',hue='gender')
plt.title('Average Evaluation Score by gender and age')
plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.show()
```

suggests that age doesnt have a stromng relationship with teaching eval scores for either male or female instructors . both genders show variation in evaluation scores across different age groups

```python
sns.scatterplot(data=df,x='age',y='eval',hue='gender',style='tenure')
plt.title('Age and Evaluation Score by gender and tenure')
plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.show()
```

it shows that teaching evaluation scores vary considerably across age groups for both genders and tenure status. there is no clear relationship between instructor age and evaluation score , after considering gender and tenure
