# Practical 5 - Understanding Two Tailed Test

Aim:

To understand probability and perform a two-tailed Z-test using Python.

Objectives:

To calculate the probability of evaluation scores greater than a given value.
To calculate the probability of evaluation scores within a specified range.
To understand null and alternative hypotheses.
To calculate the Z-score and p-value.
To interpret the results of a two-tailed hypothesis test.

```python
import pandas as pd
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

url = "https://vincentarelbundock.github.io/Rdatasets/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)
df.head()

# df[['eval']].describe()
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print("mean:",df['eval'].mean())
print("stantard deviation:",df['eval'].std())
```

```python
mean=df['eval'].mean()
std=df['eval'].std()
x=4.5
z=(x-mean)/std
# prob=P(X>4.5)
p=1-norm.cdf(z)
print("Mean=", mean)
print("standard deviation:",std)
print("probability of evaluation score>4.5=",p)
print("probability %=", p*100)
```

the probability of receiving an eval. score> 4.5 is approx 18.29%

```python
lower=3.5
upper=4.2
zlower=(lower-mean)/std
zupper=(upper-mean)/std
plower=norm.cdf(zlower)
pupper=norm.cdf(zupper)
prange=pupper-plower
print("z score lower=",zlower)
print("z score upper=",zupper)
print("probability=",prange)
print("probability %=", prange*100)
```

the probability of receiving an eval. score between 3.5 and 4.2 is approx 45%

```python
mu=12
sigma=5.5
sample_mean=10.7
n=36
alpha=0.05

stderror=sigma/np.sqrt(n)
zstat=(sample_mean-mu)/stderror
p_value=2*(1-norm.cdf(abs(zstat)))
print("standard error=",stderror)
print("z-statistic=",zstat)
print("p-value=",p_value)
```

**null hypothersis**
$H_0: \mu = 12$ the mean score of the regional players is not significantly different from the historical mean of 12 points
alternative hypothesis:
$H_1: \mu \neq 12$
the mean score of regional players is significantly different from the historical mean of 12 points

```python
criticalval=norm.ppf(1-alpha/2)
print("critical value=",-criticalval,"and",criticalval)
if abs(zstat)>criticalval:
    print("reject null hypothesis")
else:
    print("fail to reject null hypothesis")
```

**Interpretation**

At the 5% significance level, the two-tailed test does not provide sufficient evidence to conclude that the regional players' mean scoring average differs from the historic mean of 12 points.

Decision: Fail to reject H₀.

The calculated Z-score is -1.4182 and the two-tailed p-value is 0.1561.

Since the p-value (0.1561) is greater than the significance level (0.05), we fail to reject the null hypothesis.

Therefore, there is insufficient statistical evidence to conclude that the mean points scored by regional players is different from the historic mean of 12 points.
This computes critical value, the test statistic falls in the rejection region, so we reject H₀; otherwise, we fail to reject H₀.
