# **Practical**-7 Understanding Regression with T-Test and Anova

Aim:

To study the relationship between teachers' gender, age, beauty scores, and teaching evaluation scores using regression analysis, T-test, ANOVA, and correlation.

Objectives:

To understand regression analysis and its application in statistics.
To examine whether gender affects teaching evaluation scores using a T-test.
To study whether instructors' beauty scores differ according to age groups using ANOVA.
To examine the relationship between teaching evaluation scores and beauty scores.
To understand the interpretation of regression coefficients and p-values.
To draw conclusions from statistical results using the TeachingRatings dataset.

```python
import pandas as pd
from scipy import stats
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm

url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"

ratings_df = pd.read_csv(url)

ratings_df.head()
```

**Q1. Regression with T-test**

```python
model = smf.ols("eval ~ C(gender)", data=ratings_df).fit()

print(model.summary())
```

The regression analysis shows that male professors have an average evaluation score 0.168 points higher than female professors. The p-value (0.001) is less than 0.05, indicating a statistically significant difference in evaluation scores between genders. However, the R-squared value (0.022) suggests that gender explains only 2.2% of the variation in evaluation scores.



**Q2. Regression with ANOVA**

```python
ratings_df["age_group"] = pd.cut(
    ratings_df["age"],
    bins=[-float("inf"), 40, 56, float("inf")],
    labels=["40 and younger", "41 to 56", "57 and older"]
)

model = smf.ols("beauty ~ C(age_group)", data=ratings_df).fit()

print(anova_lm(model))
```

The ANOVA test shows a statistically significant difference in beauty scores among the three age groups of professors. Since the p-value (4.32 × 10⁻⁸) is less than 0.05, we reject the null hypothesis and conclude that at least one age group has a significantly different mean beauty score. Further analysis is needed to identify which specific age groups differ..

**Q3. Correlation**

```python
model = smf.ols("eval ~ beauty", data=ratings_df).fit()

print(model.summary())
```

The regression analysis shows a positive relationship between professors’ beauty scores and their evaluation scores. For every one-unit increase in beauty score, the evaluation score increases by 0.133 on average, and the relationship is statistically significant (p-value < 0.001). However, the R-squared value of 0.036 indicates that beauty scores explain only 3.6% of the variation in evaluation scores.
