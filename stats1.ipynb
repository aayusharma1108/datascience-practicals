# Practical 1: Revisiting Pandas and NumPy

**Aim :** To revise the basic concepts of Pandas and NumPy and perform various numerical and matrix operations using Python.

**Objectives:**

To understand Pandas and NumPy libraries.
To create and manipulate a Pandas DataFrame.
To convert a DataFrame into a NumPy array.
To understand the shape, size, and dimensions of an array.
To perform basic arithmetic and matrix operations.

**Step 1**: Importing the necessary libraries. Pandas (pd) is used to create and manage DataFrames. NumPy (np) is used for numerical and matrix operations such as addition, multiplication, transpose, and inverse.

**Step 2:** Create a Dataset and DataFrame. Data stores the raw numerical values in rows and columns. pd.DataFrame() converts the raw data into a DataFrame.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Mahi", "Shikha", "Pallavi", "Akshita", "Tushar", "Anshul"],
    "Stats": [85, 78, 92, 67, 74, 88],
    "Physics": [76, 82, 89, 71, 65, 90],
    "AI": [91, 85, 80, 73, 79, 95]
}

df = pd.DataFrame(data)
print(df)
```

**Step 3:** Convert DataFrame to NumPy Array. df.to_numpy() converts the Pandas DataFrame into a NumPy array. The array is stored in the variable arr.

**Step 4:** Finding Shape, Size and Dimension. arr.shape → shows the number of rows and columns. arr.size → shows the total number of elements. arr.ndim → shows the number of dimensions.

```python
marks_arr = df[["Stats", "Physics", "AI"]].to_numpy()
print(marks_arr.shape)
print(marks_arr.size)
print(marks_arr.ndim)
print("numpy Array: ")
print(marks_arr)
```

```python
additional_arr=np.array([
    [4,6,7],
    [2,6,8],
    [5,6,9],
    [5,7,5],
    [5,7,2],
    [6,7,1]
])
addition= marks_arr + additional_arr
print("addition: ")
print(addition)

subtraction= marks_arr - additional_arr
print("\nsubtraction: ")
print(subtraction)

multiplication= marks_arr * additional_arr
print("\nmultiplication: ")
print(multiplication)

transpose= marks_arr.T
print("\ntranspose: ")
print(transpose)

matrix_product= np.dot(marks_arr, additional_arr.T)
print("\nmatrix product: ")
print(matrix_product)

inverse= np.linalg.inv(matrix_product)
print("\ninverse: ")
print(inverse)
```

## Conclusion
In this practical, we revised the basics of Pandas and NumPy. We created a DataFrame, converted it into a NumPy array, and explored its shape, size, and dimensions. We performed addition, subtraction, element-wise multiplication, transpose, and matrix multiplication.
